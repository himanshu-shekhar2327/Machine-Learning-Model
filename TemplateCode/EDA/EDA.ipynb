```python
cat_cols = ['Sex', 'Embarked','Pclass','SibSp']
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for i, col in enumerate(cat_cols):
    sns.countplot(x=col, data=titanic, ax=axes[i])
    axes[i].set_title(col)
    axes[i].tick_params(axis='x', rotation=45)

for i in range(len(cat_cols) , len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.show()
```

# Bivariate Analysis

## 1. Scatterplot (Numerical - Numerical)

```python
# Using tips dataset
sns.scatterplot(data = tips , x = tips['total_bill'] , y = tips['tip']) 
# This is bivariate between two features
```

```python
# Here each dot reperesent a customer of tips datast and it is linear
```

```python
 # Multivariate between 3 features
sns.scatterplot(data = tips , x = tips['total_bill'] , y = tips['tip'] , hue = tips['sex']) 
```

```python
# Multivariate 4 features
sns.scatterplot(data = tips , x = tips['total_bill'] , y=tips['tip'] , hue = tips['sex'] , style = tips['smoker'])
```

```python
#Multivariate 5 features
sns.scatterplot(data = tips , x = tips['total_bill'] , y=tips['tip'] , hue=tips['sex'] , style = tips['smoker'] , size = tips['size'])
```

## 2. Barplot (Numerical - Categorical) 


```python
titanic.head()
```

```python
sns.barplot(data = titanic , x = titanic['Pclass'] ,y= titanic['Fare'] , hue = titanic['Sex'])
```

## 3. Boxplot

```python
sns.boxplot(data = titanic , x = titanic['Sex'] , y=titanic['Age'] , hue = titanic['Survived'])
```

## 4. DistPlot (Numerical - Categorical)

```python
sns.kdeplot(data = titanic[titanic['Survived'] == 0]['Age'] , label='Not Survived')
sns.kdeplot(data = titanic[titanic['Survived'] == 1]['Age'] , label ='Survived')
```

```python
sns.kdeplot(data = titanic,
           x = 'Age',
           hue ='Survived',
            common_norm = False
            )
```

# HeatMap (Categorical - Categorical)

```python
pd.crosstab(titanic['Pclass'] , titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'] , titanic['Survived']))
```

```python
(titanic.groupby('Pclass').mean(numeric_only = True)['Survived']*100).plot(kind='bar')
```

```python
(titanic.groupby('Pclass').mean(numeric_only = True)['Survived']*100)
```

# ClusterMap(Categrocial - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['SibSp'], titanic['Survived']))
```

# Pairplot

```python
sns.pairplot(data = titanic , hue='Survived')
```
