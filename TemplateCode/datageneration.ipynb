```python
# ===== Template: Generate and visualize circular data =====
from sklearn.datasets import make_circles
import matplotlib.pyplot as plt

def generate_circle_data(samples=100, factor=0.5, noise=0.1):
    X, y = make_circles(n_samples=samples, factor=factor, noise=noise, random_state=42)
    plt.figure(figsize=(6,5))
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k')
    plt.title('Circular Dataset')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.show()
    return X, y
X, y = generate_circle_data(200, 0.4, 0.15)
```

```python
# make_circles is only for 2D circular patterns, so it doesn’t support n_features > 2.

# But sklearn gives you other generators that do:

# 🔹 make_classification

# This lets you control:

# Number of features

# Number of informative features

# Redundant features

# Noise

# Classes

from sklearn.datasets import make_classification
import matplotlib.pyplot as plt

X, y = make_classification(
    n_samples=200,        # number of data points
    n_features=5,         # number of features (columns)
    n_informative=3,      # how many features actually affect the label
    n_redundant=1,        # how many are duplicates or linear combinations
    n_classes=2,          # binary classification
    random_state=42
)

print(X.shape)
# Output: (200, 5)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
# → Even though your dataset has 5 features, you just plot 2 of them for visualization.
```

```python
# add artificial features yourself
from sklearn.datasets import make_circles
import numpy as np

X, y = make_circles(n_samples=100, factor=0.5, noise=0.1, random_state=0)

# add 3 random features (noise)
extra_features = np.random.randn(100, 3)
X_new = np.hstack([X, extra_features])

print(X_new.shape)
# (100, 5)  --> 2 from make_circles + 3 new
```
